```python
import tensorflow as tf
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import Dense, Flatten 
from tensorflow.keras.datasets import mnist #datasets
from tensorflow.keras.utils import to_categorical # this is use for one hot encoding
```

```python
(x_train,y_train),(x_test,y_test) = mnist.load_data()
```

```python
x_train = x_train/255.0
x_test = x_test/255.0
```

```python
y_train = to_categorical(y_train,10)
y_test = to_categorical(y_test,10)
```

```python
modal = Sequential([
 Flatten(input_shape=(28, 28)), # 28x28 images to 784 input features
    
 Dense(128, activation='relu'), # Hidden layer with 128 neurons 
 Dense(10, activation='softmax') # Output layer for 10 classes
])
```

```python
modal.compile( 
    optimizer='adam',
    loss='categorical_crossentropy', 
    metrics=['accuracy']
)
```

```python
modal.fit(x_train,y_train,epochs= 5,batch_size = 32)
```

```python
test_loss,test_acc = modal.evaluate(x_test,y_test)
print(f"Test accuracyis {test_acc}")
```
