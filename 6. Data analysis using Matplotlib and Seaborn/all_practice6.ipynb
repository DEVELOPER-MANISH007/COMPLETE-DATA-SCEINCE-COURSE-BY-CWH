### 1. Basic Scatter Plot
- Scatter plots are used to show the relationship between two variables. Let's assume we are plotting the study
hours vs. exam scores for a group of students.

````python
import matplotlib.pyplot as plt
# Sample data
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 10]
plt.scatter(study_hours, exam_scores)
plt.title('Study Hours vs Exam Score')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
````

```Note:```This shows a simple scatter plot. You can notice a general upward trend — more study hours lead to
better scores

### 2. Adding Color and Size

````python
# Size of points based on score (bigger score -> bigger point)
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 90, 50, 5, 60, 65, 75, 85, 10]
size = [score*2 for score in exam_scores]
colors = ['red' if score>60 else 'green' for score in exam_scores]
plt.scatter(study_hours,exam_scores, s=  size,  c = colors)
plt.title('Colored & sized Scatter Plot')
plt.xlabel('Exam Score')
plt.grid(True)
plt.show()
````

### 3. Using a Colormap|

````python
import matplotlib.pyplot as plt
study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 90]
color= [12,23,34,46,56,76,45,45,34]
plt.scatter (study_hours, exam_scores, c = color, cmap= 'viridis')
plt.colorbar(label = 'Score')
plt.xlabel('Study Hours')
plt.ylabel('Student Marks')
plt.title('Exam Score')
plt.grid(True)
````

> Google: https://matplotlib.org/stable/users/explain/colors/colormaps.html
   - cmap adds gradient coloring based on the score. colorbar helps understand what the colors
    represent.


| Parameter | Purpose | Example | Important Note |
|-----------|---------|---------|----------------|
| `s`       | Marker size (area in points²) | `s=100` or `s=[50,100,200]` | Bigger value → bigger circle |
| `c`       | Marker color values | `c="red"` or `c=[10,20,30]` | Can be single color or numeric array |
| `cmap`    | Color gradient mapping | `cmap="viridis"` | Works when `c` is numeric |
| `size`    | ❌ Not valid parameter | — | Use `s` instead |

| `c` Type            | `cmap` Required?                        | Result           |
| ------------------- | --------------------------------------- | ---------------- |
| Numeric values      | ❌ Not required (default apply hota hai) | Gradient color   |
| Numeric values      | ✅ If given                              | Custom gradient  |
| Color names         | ❌                                       | Direct color use |
| Single color string | ❌                                       | Sab same color   |


## 4. Adding Annotations

````python
plt.scatter(study_hours, exam_scores,c= [12,23,34,45,56,67,78,89,90] , s = [100,200,300,400,500,600,700,800,900])
# add labels
for i in range(len(study_hours)):
    plt.annotate(f'S{i+1}', (study_hours[i], exam_scores[i]), )
plt.title('Scatterr Plot with Aannotations')
plt.xlabel('Study Hours')
plt.colorbar(label = 'color')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
````

```Annotations help identify individual points, which is useful in small datasets.```

## 5. Multiple Groups in One Plot

````python
# Assume two groups: Class A and Class B
class_a_hours = [2, 4, 6, 8]
class_a_scores = [45, 55, 65, 85]
class_b_hours = [1, 3, 5, 7, 9]
class_b_scores = [40, 50, 60, 70, 90]
plt.scatter(class_a_hours, class_a_scores, label = 'Class A' , color = 'red')
plt.scatter(class_b_hours, class_b_scores, label = 'Class B', color = 'green')
plt.title('Scatter plot:Class A vs Class B')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.legend()
plt.show()
````

```Note:```When comparing two datasets, use different colors and a legend for clarity
