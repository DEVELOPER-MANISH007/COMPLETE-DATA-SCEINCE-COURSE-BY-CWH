###  Matplotlib Pie Charts
- Pie charts are used to show ```part-to-whole``` relationships. They're visually appealing but best used with ```fewer
categories,``` as too many slices can get cluttered and hard to interpret.


#### basic Pie Chart Example

````python
#  Data
import matplotlib.pyplot as plt
labels = ["Sachin","Sehwag","Kohli","Yuvraj"]
runs = [18000,8000,12000,9500]
plt.title("Caraeer Runs Of Indian BatsMan")
plt.pie(runs,labels = labels)
plt.show()
````

### Custom Colors

````python
colors = ['red','green','yellow','black']
plt.pie(runs, labels=labels, colors=colors)
plt.show()
````

### 📌 Markdown Table – wedgeprops Parameters

| Property      | Purpose                                      | Example Value |
|---------------|----------------------------------------------|--------------|
| edgecolor     | Slice ke border ka color                    | "black"     |
| linewidth     | Border ki thickness                         | 2           |
| linestyle     | Border ka pattern (-, --, :, -.)            | "--"        |
| width         | Donut chart ke liye slice thickness         | 0.4         |
| alpha         | Transparency level (0 to 1)                 | 0.8         |
| hatch         | Pattern fill (/, \\, |, -, +, x, o, etc.)   | "///"       |
| facecolor     | Slice ka fill color                         | "red"       |
| antialiased   | Smooth edges enable/disable                 | True        |
| zorder        | Layer order control karta hai               | 2           |
| capstyle      | Border end style (butt, round, projecting)  | "round"     |
| joinstyle     | Border join style (miter, round, bevel)     | "miter"     |

### Add Edges and Style Slices with wedgeprops
- You can customize the look of the slices using wedgeprops

````python
plt.pie(
 runs,
 labels=labels,
 colors=colors,
 wedgeprops={'edgecolor': 'yellow', 'linewidth': 2, 'linestyle': '-', }
)
plt.show()
````

### Highlight a Slice Using explode
- Use ```explode``` to ```pull out``` one or more slices.


````python
explode = [0.099, 0, 0, 0] # Only highlight Sachin's slice
plt.pie(runs, labels=labels, explode=explode, colors=colors)
plt.title("Exploded Pie Example")
plt.show()
````

### Add Shadows for a 3D Feel

````python
plt.pie(
 runs,
 labels=labels,
 explode=explode,
 # colors=colors,
 shadow=True # adds a 3D-like shadow
)
plt.show()
````

### Start Angle
- Use startangle to rotate the pie for better alignment

````python
explode = [0.1, 0, 0, 0] # Only highlight Sachin's slice
plt.pie(
    runs,
    labels = labels,
    explode = explode,
    shadow = True,
    startangle = 140)       # change angle 
plt.show()
````

### Show Percentages with autopct

````python
plt.pie( runs,labels = labels, autopct= '%1.1f%%',  startangle = 140,
       wedgeprops= {'edgecolor':'yellow','linewidth':5, 'linestyle':'-'}
       )
plt.show()
````

### ```Crowded Pie Chart – Why to Avoid```

````python
# Too many categories
languages = ['Python', 'Java', 'C++', 'JavaScript', 'C#', 'Ruby', 'Go', 'Rust',
'Swift', 'PHP']
usage = [30, 20, 10, 10, 7, 5, 4, 3, 2, 1]
plt.pie(usage,labels = languages, autopct = '%1.1f%%',  startangle = 90)
plt.title("Programming Language Usage (Crowded Example)")
plt.show()
````

### Summary of Useful Pie Chart Parameters

| Parameter    | Use |
|:-------------|:----|
| `labels`     | Adds labels to each slice |
| `colors`     | Customizes slice colors |
| `explode`    | Highlights slices by pulling them outward |
| `shadow`     | Adds a depth-like shadow effect |
| `startangle` | Rotates the pie chart |
| `autopct`    | Displays percentage values |
| `wedgeprops` | Controls slice border, width, and styling |

Examples:

````python
import matplotlib.pyplot as plt
Values = [30,25,20,15,10]
labels = ['Python','Java','C++','JavaScripts','Go']
plt.pie(Values,labels = labels,  autopct = '%1.1f%%' ,wedgeprops ={
    'linestyle':'-', 'edgecolor':'black' 
})
plt.title("Languages with its percentage")
````

````python
import matplotlib.pyplot as plt

Values = [30,25,20,15,10]
colors = ['yellow','red', 'green','blue','orange']
labels = ['Python','Java','C++','JavaScript','Go']

plt.xlabel('Values')
plt.ylabel('Languages')
plt.title("Languages with its percentage")

plt.barh(labels, Values, color=colors, label="Language Popularity")  # 👈 label add kiya
plt.legend()

plt.show()
````
