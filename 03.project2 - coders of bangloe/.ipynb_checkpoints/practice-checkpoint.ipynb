# Data Collection
- You hire Vijyalaxmi Iyer and Sam Chandra who are from Hebbal and HSR Layout respectively
- HSR layout and Hebbal are far away so you set up a meeting at Rameshwaram cafe, Indiranagar

```python
with open("06_finaldata.txt", encoding='utf-8') as f:
    data = f.read()
chunks = data.split("\n\n") 
    
def parse_chunk(chunk):
    chunk = chunk.strip()
    # skip empty chunks
    if not chunk:
        return None
    sep_chunk = chunk.split('\n')
    if len(sep_chunk)<6:
        return None
    username = sep_chunk[0]
    no_of_posts = int (sep_chunk[1].split(" post")[0].replace(",",""))
    no_of_followers = sep_chunk[2].split(" followers")[0].replace(",","").replace("K","000").replace("M","000000")
    raw_followers = sep_chunk[2].split(" followers")[0].replace(",", "").strip()
    if raw_followers.lower().endswith("k"):
        no_of_followers = int(float(raw_followers[:-1]) * 1000)
    elif raw_followers.lower().endswith("m"):
        no_of_followers = int(float(raw_followers[:-1]) * 1000000)
    else:
        no_of_followers = int(raw_followers)
    no_of_following = int (sep_chunk[3].split(" following")[0].replace(",",""))
    name = sep_chunk[4]
    account_type = sep_chunk[5]
    bio = "\n".join(sep_chunk[6:])
    
    # print(username, no_of_posts, no_of_following, name, account_type, bio, sep = "\n"
    
    return {
    "username": username,
    "no_of_posts": no_of_posts,
    "no_of_followers":no_of_followers,
    "no_of_following": no_of_following,
    "name": name,
    "account_type": account_type,
    "bio": bio
}

all_chunks = []
for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    if parsed_chunk:
        all_chunks.append(parsed_chunk)
print(all_chunks)

import json
s = json.dumps(all_chunks, indent = 4)
print(s)
with open('data.json','w' ) as f:
     f.write(s)
    
```
