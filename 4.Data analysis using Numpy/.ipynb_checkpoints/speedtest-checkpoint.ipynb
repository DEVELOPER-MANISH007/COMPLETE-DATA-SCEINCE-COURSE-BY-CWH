# why use Numpy

- bhai saaf sidhi baat hai python ki list slow hoty hai camparison kre to numpy se , numpy jyda fast hota hai bhai

## lets check it 

```python
# add list using python list

import numpy as np
import time 

size = 10000
l1 =  list(range(size))
l2 = list(range(size))

start = time.time()
add = [x+y for x,y in zip(l1,l2)]
end = time.time()
print(end-start)
```

#### Add list uing numpy

```python
import numpy as np
import time 

size = 10000
l1 =  np.array(list(range(size)))
l2 = np.array(list(range(size)))
start = time.time()
add = l1+l2
end = time.time()
print(end-start)

```

### conclusion ye hai ki time ko madde nazar dekhte hue numpy fast hai
